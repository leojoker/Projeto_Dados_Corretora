# 📈 Análise de Indicadores de Canais de Marketing
Este notebook apresenta uma análise dos canais de marketing com base em:
- Total de leads
- Taxa de conversão
- Depósito médio

O objetivo é identificar os canais com melhor performance e maior potencial de investimento.

````python
import pandas as pd
import plotly.express as px

# Carregar dados
df = pd.read_csv("D:/pythonProject/Projeto_Dados_Corretora/Data/FINAL/indicadores_por_canal.csv")
df.head()
````

## 🔹 Total de Leads por Canal

````python
fig = px.bar(df.sort_values("total_leads", ascending=False), 
             x="total_leads", y="origem", orientation="h",
             title="Total de Leads por Canal",
             text_auto=True, color="total_leads", color_continuous_scale="Blues")
fig.update_layout(yaxis_title="", xaxis_title="Total de Leads", height=400)
fig.show()
````

## 🔹 Taxa de Conversão por Canal

````python
fig = px.bar(df.sort_values("taxa_conversao", ascending=True), 
             x="taxa_conversao", y="origem", orientation="h",
             title="Taxa de Conversão (%) por Canal",
             text_auto=".2%", color="taxa_conversao", color_continuous_scale="Aggrnyl")
fig.update_layout(yaxis_title="", xaxis_title="Taxa de Conversão", height=400)
fig.show()
````

## 🔹 Depósito Médio por Canal

````python
fig = px.bar(df.sort_values("deposito_medio", ascending=True), 
             x="deposito_medio", y="origem", orientation="h",
             title="Valor Médio de Depósito por Canal",
             text_auto=".2f", color="deposito_medio", color_continuous_scale="YlOrBr")
fig.update_layout(yaxis_title="", xaxis_title="Valor Médio (R$)", height=400)
fig.show()
````

# 🎯 Estratégias para Ajustar os Pesos

Abaixo estão algumas estratégias para definir os pesos da métrica `score_combinado`, de acordo com o objetivo do projeto de marketing:

📘 [Documentação: Estratégia de Score Combinado](D:/pythonProject/Projeto_Dados_Corretora/Docs/score_combinado.md)


---

## 1. **Foco em Eficiência (Performance)**
Você quer priorizar canais que **convertem melhor**:

```python
taxa_conversao = 0.6  
deposito_medio = 0.3  
total_leads = 0.1
```

---

## 2. **Foco em Receita (Ticket Médio)**
Você quer os canais com **maior retorno financeiro médio**:

```python
taxa_conversao = 0.3  
deposito_medio = 0.5  
total_leads = 0.2
```

---

## 3. **Foco em Volume (Escalabilidade)**
Você quer **alcançar o maior número de pessoas possível**:

```python
taxa_conversao = 0.2  
deposito_medio = 0.2  
total_leads = 0.6
```

---

## 4. **Foco Balanceado (Padrão Atual)**
Esse é o que usamos neste notebook:

```python
taxa_conversao = 0.5  
deposito_medio = 0.3  
total_leads = 0.2
```

---

````python
from IPython.display import Markdown

# Cria uma métrica de score combinado (ajuste os pesos conforme o contexto)
df['score_combinado'] = (
    df['taxa_conversao'] * 0.5 +
    (df['deposito_medio'] / df['deposito_medio'].max()) * 0.3 +
    (df['total_leads'] / df['total_leads'].max()) * 0.2
)

# Ordena do melhor para o pior canal
df_sorted = df.sort_values(by='score_combinado', ascending=False)

# Geração do markdown dinâmico
md_texto = "# 📊 Interpretação Estratégica dos Canais de Marketing\n\n"

for _, row in df_sorted.iterrows():
    canal = row['origem']
    conv = row['taxa_conversao']
    dep = row['deposito_medio']
    vol = row['total_leads']
    score = row['score_combinado']

    emoji = "✅"
    comentario = ""

    if conv > 0.052 and dep > 260:
        comentario = "Alta taxa de conversão e alto ticket médio. Merece mais investimento 🔥"
    elif conv > 0.05:
        comentario = "Boa taxa de conversão. Potencial para escalar 🚀"
    elif conv < 0.049:
        comentario = "Conversão baixa. Precisa de otimização ⚠️"
    else:
        comentario = "Canal confiável com desempenho estável. Ideal para manter 💡"

    md_texto += f"### {emoji} {canal}\n"
    md_texto += f"- 📈 Taxa de Conversão: **{conv:.2%}**\n"
    md_texto += f"- 💰 Depósito Médio: **R${dep:,.2f}**\n"
    md_texto += f"- 👥 Volume de Leads: **{vol:,}**\n"
    md_texto += f"- 🧮 Score Combinado: **{score:.3f}**\n"
    md_texto += f"- 📌 {comentario}\n\n---\n"

# Exibe o markdown no notebook
Markdown(md_texto)
````
