# 📦 Geração de Leads Semanais Simulados (com BASE_DIR)

Geração de arquivos semanais salvos na estrutura `Data/RAW/FULL_SIMULATED/{ano}/{mes}`.

```python
import pandas as pd
import random
from faker import Faker
from datetime import datetime, timedelta
from pathlib import Path
import numpy as np
import sys

# Adiciona caminho raiz para importações
sys.path.append(str(Path.cwd().parent))
from Utils.io import save_csv

# Define caminhos base
BASE_DIR = Path.cwd().parent
raw_dir = BASE_DIR / 'Data' / 'RAW' / 'FULL_SIMULATED'

# Inicializa Faker
fake = Faker()

# Datas de início e fim
start_date = datetime(2023, 4, 1)
end_date = datetime(2025, 4, 20)

# Gera lista de semanas
semanas = pd.date_range(start=start_date, end=end_date, freq='W-MON')
num_semanas = len(semanas)

# Leads com crescimento progressivo
leads_por_semana = np.linspace(1000, 4000, num=num_semanas).astype(int)

lead_id = 1

# Loop semana a semana
for i, data_ref in enumerate(semanas):
    ano = data_ref.year
    mes = f"{data_ref.month:02d}"
    semana_mes = (data_ref.day - 1) // 7 + 1
    semana_str = f"{semana_mes:02d}"
    qtd = leads_por_semana[i]

    dados_semana = []
    
    for _ in range(qtd):
        data_cadastro = fake.date_between(start_date=data_ref, end_date=data_ref + timedelta(days=6))
        convertido = random.choices(['Convertido', 'Não Convertido'], weights=[0.05, 0.95])[0]
        dias_trade = random.randint(0, 30) if convertido == 'Convertido' else None
        valor_deposito = round(random.uniform(50, 10000), 2) if convertido == 'Convertido' else 0.0

        dados_semana.append({
            'lead_id': lead_id,
            'origem': random.choices(
                ['Orgânico', 'Instagram', 'Facebook', 'TikTok', 'E-mail Marketing', 'Google Ads', 'YouTube Ads', 'WhatsApp', 'Indicação'],
                weights=[0.18, 0.15, 0.10, 0.14, 0.12, 0.10, 0.08, 0.08, 0.05]
            )[0],
            'data_cadastro': data_cadastro,
            'status_conversao': convertido,
            'dias_ate_1o_trade': dias_trade,
            'valor_deposito': valor_deposito,
            'perfil': random.choices(
                ['Conservador', 'Moderado', 'Agressivo', 'Desconhecido'],
                weights=[0.2, 0.4, 0.3, 0.1]
            )[0],
            'pais': fake.country()
        })
        lead_id += 1

    df_semana = pd.DataFrame(dados_semana)

    # Pasta do mês
    output_dir = raw_dir / str(ano) / mes
    output_dir.mkdir(parents=True, exist_ok=True)

    nome_arquivo = f"{ano}_{mes}_{semana_str}.csv"
    df_semana.to_csv(output_dir / nome_arquivo, index=False)
    print(f"✅ {nome_arquivo} salvo com {len(df_semana)} leads")
```
